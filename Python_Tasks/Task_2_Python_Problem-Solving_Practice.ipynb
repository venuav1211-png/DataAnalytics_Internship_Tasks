# Q1. Find the Shortest Word in a Sentence
Write a Python program to find the shortest word in a given sentence.

Example input:

Python makes coding easy

Expected output:

easy


```python
sentence = "Python makes coding easy"
words = sentence.split()

shortest = words[0]
for word in words:
    if len(word) < len(shortest):
        shortest = word

print(shortest)  
```

# Q6. Count Vowels in a String
Write a Python program to count the total number of vowels in a string. Consider both uppercase and lowercase vowels.

Example input:

Education

Expected output:

5


```python
s = "Education"
vowels = "aeiouAEIOU"
count = 0

for char in s:
    if char in vowels:
        count += 1

print(count) 
```

# Q7. Count Consonants in a String
Write a Python program to count the consonants in a string. Ignore spaces, digits, and special characters.

Example input:

Python 123!

Expected output:

5


```python
s = "Python 123!"
vowels = "aeiouAEIOU"
count = 0

for char in s:
    if char.isalpha() and char not in vowels:
        count += 1

print(count) 
```

# Q8. Find the Most Frequent Character
Write a Python program to find the character that appears most frequently in a string. 
If multiple characters have the same highest frequency, return the first one encountered. Ignore spaces.

Example input:

banana

Expected output:

a


```python
s = "banana"
s = s.replace(" ", "")

freq = {}
for char in s:
    freq[char] = freq.get(char, 0) + 1

max_char = s[0]
for char in s:
    if freq[char] > freq[max_char]:
        max_char = char

print(max_char)  
```

# Q10. Find All Characters Appearing More Than Once
Write a Python program to identify all characters that appear more than once in a string. Display each repeated character only once, in the order of its first appearance. Ignore spaces.

Example input:

programming

Expected output:

['r', 'g', 'm']


```python
s = "programming"
s = s.replace(" ", "")

freq = {}
for char in s:
    freq[char] = freq.get(char, 0) + 1

result = []
for char in s:
    if freq[char] > 1 and char not in result:
        result.append(char)

print(result) 
```

# Q15. Create a Dictionary of Numbers and Their Even/Odd Status
Write a Python program to create a dictionary containing each number as a key and "Even" or "Odd" as its value.

Example input:

[1, 2, 3, 4, 5]

Expected output:

{1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}


```python
nums = [1, 2, 3, 4, 5]

result = {n: ("Even" if n % 2 == 0 else "Odd") for n in nums}
print(result) 
```

# Q16. Find the Most Frequent Word in a Sentence

# Q20. Find the Longest Word That Appears Only Once
Write a Python program to find the longest word that appears exactly once in a sentence. If multiple qualifying words have the same maximum length, return the first one encountered. Treat uppercase and lowercase versions as the same word.

Example input:

Python makes coding useful Python

Expected output:

coding


```python
sentence = "Python makes coding useful Python"
words = sentence.lower().split()

freq = {}
for word in words:
    freq[word] = freq.get(word, 0) + 1

longest = ""
for word in words:
    if freq[word] == 1 and len(word) > len(longest):
        longest = word

print(longest)
```
